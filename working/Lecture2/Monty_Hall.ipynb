```python
import numpy as np
import random as ran
import matplotlib.pyplot as plt
```

```python
# giocatore_c contiene il valore della porta selezionata, ovvero 0,1,2

swi_c = 0    #giocatore che cambierà porta
swi = 0

con_c = 0    #giocatore che tiene la sua porta
con = 0

new_c = 0    #giocatore che arriva con la porta aperta
new = 0

n = 1000

doors = [0, 0, 1]

cond = 0   #conduttore
```

```python
for i in range (n):
    
    con = ran.randrange(0, 3, 1)

    cond = ran.choice([x for x in range(0,2) if x != con])

    swi = ran.choice([x for x in range(0,3) if (x != con) & (x != cond)])

    new = ran.choice([x for x in range(0,3) if x != cond])

    if doors[con] == 1: con_c += 1
    if doors[swi] == 1: swi_c += 1
    if doors[new] == 1: new_c += 1
                  

    #A questo punto ho il conduttore che ha aperto una porta con lo zero diversa dalla porta che ha scelto il conduttore conservativo

    
```

```python
print(con_c)
print(swi_c)
print(new_c)
```

```python
labels = ["conservative", "switcher", "new comer"]
results = [con_c, swi_c, new_c]
bins = [0, 1, 2]
```

```python
plt.bar(labels, results, width = 1)
plt.ylabel("Probability")
plt.xlabel("Players")
plt.show()
```
